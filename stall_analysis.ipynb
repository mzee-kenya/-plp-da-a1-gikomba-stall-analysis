```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
# Calculate total and average sales
total = sum(daily_sales)
average_sales = total / len(daily_sales)

# Identify highest and lowest sales
highest_sale = max(daily_sales)
lowest_sale = min(daily_sales)

# Find the positions of highest and lowest sales
best_position = daily_sales.index(highest_sale)
worst_position = daily_sales.index(lowest_sale)

# Identify the corresponding days
best_day = days[best_position]
worst_day = days[worst_position]

# Determine the week numbers
best_week = 1 if best_position < 7 else 2
worst_week = 1 if worst_position < 7 else 2

# Print the sales performance results
print("OVERALL SALES PERFORMANCE")
print("-" * 40)
print(f"Total sales: KES {total:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sale: KES {highest_sale:,.2f}")
print(f"Lowest sale: KES {lowest_sale:,.2f}")
print(f"Best day: {best_day} (Week {best_week})")
print(f"Worst day: {worst_day} (Week {worst_week})")
```

```python
def week_totals(sales):
    """
    Calculate total sales for the first and second weeks.

    Args:
        sales (list): Daily sales figures.

    Returns:
        tuple: Week 1 total and Week 2 total.
    """
    week1 = sales[:7]
    week2 = sales[7:]

    return sum(week1), sum(week2)


week1_total, week2_total = week_totals(daily_sales)

growth = ((week2_total - week1_total) / week1_total) * 100

print(f"Week 1 total: KES {week1_total:,.2f}")
print(f"Week 2 total: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth:.2f}%")
```

```python
def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal or Weak.

    Args:
        amount (float): Sales amount for one day.
        average (float): Average daily sales.

    Returns:
        str: Sales performance category.
    """
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"


# Initialize a dictionary to count each category
day_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Apply the classification function to every day
for amount in daily_sales:
    category = classify_day(amount, average_sales)
    day_counts[category] += 1

# Display the results
print("DAILY SALES CLASSIFICATION")
print("-" * 40)

for category, count in day_counts.items():
    print(f"{category} days: {count}")
```

```python
# Calculate average weekday sales for both weeks
weekday_sales = daily_sales[:5] + daily_sales[7:12]
avg_weekdays_sales = sum(weekday_sales) / len(weekday_sales)

# Calculate average weekend sales for both weeks
weekend_sales = daily_sales[5:7] + daily_sales[12:14]
avg_weekends_sales = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (avg_weekends_sales - avg_weekdays_sales)
    / avg_weekdays_sales
) * 100

# Display results
print("WEEKDAY VS WEEKEND SALES")
print("-" * 40)
print(f"Average weekday sales: KES {avg_weekdays_sales:,.2f}")
print(f"Average weekend sales: KES {avg_weekends_sales:,.2f}")
print(f"Percentage difference: {percentage_difference:.2f}%")

if percentage_difference > 0:
    print("Weekend sales perform better.")
elif percentage_difference < 0:
    print("Weekday sales perform better.")
else:
    print("Weekday and weekend sales are equal.")
```

```python
# Calculate stock products that need reordering
reorder_products = []

for product, details in stock.items():
    if details["qty"] <= details["reorder"]:
        reorder_products.append(product)

# Calculate gross margin for each product
gross_margins = {}

for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    gross_margins[product] = margin


# Print the complete Stall Performance Report
print("=" * 55)
print("             STALL PERFORMANCE REPORT")
print("                 GIKOMBA MARKET")
print("=" * 55)

print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 55)
print(f"Total sales:             KES {total:,.2f}")
print(f"Average daily sales:     KES {average_sales:,.2f}")
print(f"Highest daily sale:      KES {highest_sale:,.2f}")
print(f"Lowest daily sale:       KES {lowest_sale:,.2f}")
print(f"Best sales day:          {best_day} (Week {best_week})")
print(f"Worst sales day:         {worst_day} (Week {worst_week})")

print("\n2. WEEKLY SALES PERFORMANCE")
print("-" * 55)
print(f"Week 1 sales:            KES {week1_total:,.2f}")
print(f"Week 2 sales:            KES {week2_total:,.2f}")
print(f"Weekly sales growth:     {growth:.2f}%")

print("\n3. DAILY SALES CLASSIFICATION")
print("-" * 55)
for category, count in day_counts.items():
    print(f"{category} days:             {count}")

print("\n4. WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 55)
print(f"Average weekday sales:  KES {avg_weekdays_sales:,.2f}")
print(f"Average weekend sales:  KES {avg_weekends_sales:,.2f}")
print(f"Percentage difference:  {percentage_difference:.2f}%")

print("\n5. STOCK AND GROSS MARGIN ANALYSIS")
print("-" * 55)

print("Products requiring restocking:")
for product in reorder_products:
    print(f"- {product} ({stock[product]['qty']} kg in stock)")

print("\nGross margins:")
for product, margin in gross_margins.items():
    print(f"{product:<15} {margin:>7.2f}%")

print("\n" + "=" * 55)
print("              END OF REPORT")
print("=" * 55)
```
